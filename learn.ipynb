```python
import pandas as pd
from sklearn.datasets import make_classification

X, y = make_classification(
    n_samples=1000,
    n_features=4,
    n_informative=2,
    n_redundant=1,
    weights=[0.70, 0.30],
    flip_y=0.03,
    class_sep=0.8,
    random_state=42
)

df = pd.DataFrame(
    X,
    columns=["age_score", "income_score",
             "credit_score", "experience_score"]
)

df["target"] = y
sns.heatmap(df.corr())
```

```python
import matplotlib.pyplot as plt
from sklearn.datasets import make_classification

X, y = make_classification(
    n_samples=300,
    n_features=2,
    n_informative=2,
    n_redundant=0,
    n_repeated=0,
    n_classes=3,
    n_clusters_per_class=1,
    class_sep=1,
    flip_y=0,
    random_state=42
)

plt.scatter(X[:, 0], X[:, 1], c=y, alpha=0.7)
plt.xlabel("Feature 1")
plt.ylabel("Feature 2")
plt.title("Two Clearly Separated Classes")
plt.show()
```
